```python
import pandas as pd
```

```python
df=pd.read_csv('Mobile phone price.csv')
df.head()
```

```python
missing_val=df.isnull().sum()
print(missing_val)
dup_rows=df.duplicated().sum()
print(dup_rows)
print('Shape of the data before removing duplicate rows:',df.shape)
if dup_rows>0:
  df.drop_duplicates(inplace=True)
  print('duplicate rows removed')
df.shape
```

```python
x=df.drop('Price ($)',axis=1)
y=df['Price ($)']

#encoding categorical columns
categorical_cols=x.select_dtypes(include=['object']).columns
encoded_cols=pd.get_dummies(x, columns=categorical_cols, drop_first=True)
encoded_cols=encoded_cols.astype(int)
print(encoded_cols.head())
```

```python
#Used StandardScaler from sklearn.preprocessing.
from sklearn.preprocessing import StandardScaler
#Numeric columns
numerical_columns=["Battery Capacity (mAh)"]
scaler=StandardScaler()
#Standard Scaling on numeric columns.
encoded_cols[numerical_columns]=scaler.fit_transform(encoded_cols[numerical_columns])
print(encoded_cols.head())
```

```python
#Splitting the dataset into training and testing sets
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(encoded_cols, y, test_size=0.2, random_state=42)

#Print the shape of X_train, X_test, y_train, and y_test
print("Shape of X_train:", X_train.shape)
print("Shape of X_test:", X_test.shape)
print("Shape of y_train:", y_train.shape)
print("Shape of y_test:", y_test.shape)
```

```python
#imports the Logistic Regression model from the sklearn library
from sklearn.linear_model import LogisticRegression
# Create the Linear Regression model
model= LogisticRegression()

# Fit the model on training data
model.fit(X_train, y_train)
print("Logistic Regression model trained successfully!")
```

```python
#Uses the trained model to make predictions on the test data
y_pred = model.predict(X_test)

#Print the first 10 actual values (y_test) and predicted values (y_pred).
print("First 10 Actual values of y_test:", y_test.head(10).values)
print("First 10 Predicted values of y_pred:", y_pred[:10])
```

```python

# Convert both to numeric by stripping $, commas, and spaces
y_test_clean = pd.Series(y_test).replace('[\$,]', '', regex=True).astype(float)
y_pred_clean = pd.Series(y_pred).replace('[\$,]', '', regex=True).astype(float)
```

```python
from sklearn.metrics import r2_score
r2=r2_score(y_test_clean,y_pred_clean)
print("R2 score:",r2)
```

```python
import joblib
joblib.dump(model,'model.pkl')
joblib.dump(scaler,'scaler.pkl')
joblib.dump(encoded_cols.columns,'columns.pkl')
```
